```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL da pasta no GitHub (interface web)
url_pasta = "https://github.com/FranciscoFoz/7_Days_of_Code_Alura-Python-Pandas/tree/main/Dia_1-Importando_dados/Datasets/dados_emprestimos"

# Base do link RAW (onde o pandas consegue acessar)
base_raw = "https://raw.githubusercontent.com/FranciscoFoz/7_Days_of_Code_Alura-Python-Pandas/main/Dia_1-Importando_dados/Datasets/dados_emprestimos/"

# Faz o request da página do GitHub
resposta = requests.get(url_pasta)
sopa = BeautifulSoup(resposta.text, 'html.parser')

# Filtra os nomes dos arquivos .csv
csvs = [tag.text for tag in sopa.find_all('a') if tag.text.endswith('.csv')]

# Lê e concatena todos os arquivos CSV encontrados
df_total = pd.concat([pd.read_csv(base_raw + nome) for nome in csvs], ignore_index=True)

# Pronto! Mostra o resultado
print(f"{len(csvs)} arquivos CSV carregados.")
print(df_total.head())
```
